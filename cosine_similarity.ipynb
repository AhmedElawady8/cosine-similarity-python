#**Cosine Similarity and Cosine Distance**

### **1. Cosine Similarity (cosine_similarity)**

Cosine similarity measures how similar two vectors are by
calculating the cosine of the angle between them, regardless
of their magnitude.

In NLP, it is used to measure semantic similarity between
words, sentences, or documents based on their vector
representations.

#### ***Mathematical Formula***

$$
\text{Cosine Similarity}(A,B) =
\cos(\theta) =
\frac{A \cdot B}{\|A\| \|B\|}
$$

Expanded formula:

$$
\text{Cosine Similarity}(A,B) =
\frac{\sum_{i=1}^{n} A_i B_i}
{\sqrt{\sum_{i=1}^{n} A_i^2}
\sqrt{\sum_{i=1}^{n} B_i^2}}
$$

Where:

- $A, B$: The two vectors.
- $A \cdot B$: The dot product of the two vectors.
- $\|A\|$: The magnitude of vector A.
- $\|B\|$: The magnitude of vector B.
- $\theta$: The angle between the two vectors.

#### Range: [-1, 1]

- 1: Maximum similarity (same direction).
- 0: Orthogonal vectors (no directional similarity).
- -1: Opposite directions.

---

### **2. Cosine Distance (cosine_distances)**

Cosine distance measures how different two vectors are
based on their directions rather than their magnitudes.

It is calculated by subtracting cosine similarity from 1.

#### Mathematical Formula

$$
\text{Cosine Distance}(A,B) =
1 - \text{Cosine Similarity}(A,B)
$$

Expanded formula:

$$
\text{Cosine Distance}(A,B) =
1 - \frac{A \cdot B}{\|A\| \|B\|}
$$

#### Range: [0, 2]

- 0: Maximum similarity (same direction).
- 1: Orthogonal vectors.
- 2: Opposite directions.

---

### 3. Comparison

| Feature | Cosine Similarity | Cosine Distance |
|---|---|---|
| Purpose | Measures similarity | Measures dissimilarity |
| Range | [-1, 1] | [0, 2] |
| Higher value | More similar | More different |
| Lower value | More different | More similar |
| Angle = 0° | 1 | 0 |
| Angle = 90° | 0 | 1 |
| Angle = 180° | -1 | 2 |
| Depends on magnitude | No | No |
| Depends on direction | Yes | Yes |

---

### 4. Summary

Cosine similarity measures how similar two vectors are,
while cosine distance measures how different they are.
Both depend on the angle between the vectors rather
than their magnitudes.

```python
# Import the cosine similarity and distance functions.
from sklearn.metrics.pairwise import cosine_similarity, cosine_distances
```

```python
# Compute cosine similarity for two vectors pointing in the same direction.
print(cosine_similarity([[3,1]],[[6,2]]))
```

```python
# Compute cosine similarity for two nearly aligned vectors.
print(cosine_similarity([[3,1]],[[5,2]]))
```

```python
# Compute cosine similarity for two perpendicular vectors.
print(cosine_similarity([[10,0]],[[0,2]]))
```

```python
# Define four sample documents about iPhone and Galaxy sales and trends.
doc1 = """
iphone sales contributed to 70% of revenue. iphone demand is increasing by 20% yoy.
the main competitor phone galaxy recorded 5% less growth compared to iphone"
"""

doc2 = """
The upside pressure on volumes for the iPhone 12 series, historical outperformance
in the July-September time period heading into launch event, and further catalysts in relation
to outperformance for iPhone 13 volumes relative to lowered investor expectations implies a
very attractive set up for the shares.
"""

doc3 = """
samsung's flagship product galaxy is able to penetrate more into asian markets compared to
iphone. galaxy is redesigned with new look that appeals young demographics. 60% of samsung revenues
are coming from galaxy phone sales
"""

doc4 = """
Samsung Electronics unveils its Galaxy S21 flagship, with modest spec improvements
and a significantly lower price point. Galaxy S21 price is lower by ~20% (much like the iPhone 12A),
which highlights Samsung's focus on boosting shipments and regaining market share.
"""
```

```python
# Build a document-by-term table with manually entered word counts.
import pandas as pd
df=pd.DataFrame([
    {"iphone":3,"galaxy":1},
    {"iphone":2,"galaxy":0},
    {"iphone":1,"galaxy":3},
    {"iphone":1,"galaxy":2}
],
index=[
    "doc1",
    "doc2",
    "doc3",
    "doc4"
])
```

```python
# Display the word counts for all four documents.
df
```

```python
# Display the word counts for doc1 through doc3.
df.loc["doc1":"doc3"]
```

```python
# Print the pairwise cosine similarity matrix for doc1 through doc3.
print(cosine_similarity(df.loc["doc1":"doc3"],df.loc["doc1":"doc3"]))
```

```python
# Print the cosine similarity between doc1 and doc3.
print(cosine_similarity(df.loc["doc1":"doc1"],df.loc["doc3":"doc3"]))
```

```python
# Print the cosine similarity between doc2 and doc3.
print(cosine_similarity(df.loc["doc2":"doc2"],df.loc["doc3":"doc3"]))
```

```python
# Display the cosine distance between doc1 and doc3.
cosine_distances(df.loc["doc1":"doc1"],df.loc["doc3":"doc3"])
```

```python
# Print the cosine distance between doc2 and doc3.
print(cosine_distances(df.loc["doc2":"doc2"],df.loc["doc3":"doc3"]))
```
